## Air Bnb Project

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
airbnb=pd.read_csv("Listings.csv",encoding="ISO-8859-1",low_memory=False)
airbnb
```

```python
airbnb.isna().sum()
```

```python
airbnb["host_since"]=pd.to_datetime(airbnb["host_since"])
```

```python
airbnb.info()
```

```python
paris=airbnb.query("city=='Paris'")
paris
```

```python
paris=airbnb.query("city=='Paris'").loc[:, ["host_since","neighbourhood","city","accommodates","price"]]
```

```python
paris.info()
```

```python
paris.describe()
```

```python
paris.query("accommodates==0")
```

```python
paris.query("accommodates==0 and price==0").count()
```

### Preparing for visualization

```python
paris_neighbourhood=(
    paris.groupby("neighbourhood").agg({"price":"mean"}).sort_values("price")
)
paris_neighbourhood
```

```python
paris_listings_accommodates=(
    paris.query("neighbourhood=='Elysee'").groupby("accommodates").agg({"price":"mean"}).sort_values("price")
)
```

```python
paris_listings_accommodates.head()
```

```python
paris_listings_over_time=(
    paris.set_index("host_since").resample("Y").agg({"neighbourhood":"count",
                                                     "price":"mean"})
)
```

```python
paris_listings_over_time.head()
```

### Visualize the data

```python
paris_neighbourhood.plot.barh();
```

```python
import seaborn as sns
paris_neighbourhood.plot.barh(
           title="Average Listing Price By Paris Neighbourhood",
           xlabel="Price Per Night (Euros)",
           ylabel="Neighbourhood",
           legend=None
      );
sns.despine()
```

```python
paris_listings_accommodates.plot.barh(
           title="Average Listing Price By Accommodation Number",
           xlabel="Price Per Night (Euros)",
           ylabel="Neighbourhood",
           legend=None
      );
sns.despine()
```

```python
paris_listings_over_time["neighbourhood"].plot(
    ylabel="New Hosts",
    title="New Airbnb Hosts in Paris Over Time"
)
sns.despine()
```

```python
paris_listings_over_time["price"].plot(
    ylabel="Average Price(Euros)",
    title="New Airbnb Price in Paris Over Time"
)
sns.despine()
```

```python
fig,ax=plt.subplots()
ax.plot(
      paris_listings_over_time.index,
      paris_listings_over_time["neighbourhood"],
      label="new Hosts",
      c="pink"
)
ax.set_ylabel("New Hosts")

ax2=ax.twinx()

ax2.plot(
    paris_listings_over_time.index,
      paris_listings_over_time["price"],
      label="new Hosts",
      c="blue"
)
ax2.set_ylim(0)

ax.set_ylabel("Average Price")

ax.set_title("2015 Regulations Lead to Fewer New Hosts, Higher Prices")
```
